```python
import pandas as pd

df = pd.read_csv("listings.csv")

print(df.shape)
df.info()
display(df.head())
```

```python
df.info()
display(df.describe().T)

missing = df.isnull().sum()
missing = pd.DataFrame({"jumlah": missing, "persen": (missing / len(df) * 100).round(2)})
display(missing[missing.jumlah > 0].sort_values("persen", ascending=False))
print("Duplikat:", df.duplicated().sum())
```

```python
import pandas as pd
import numpy as np

# Membuat salinan dari dataframe asli agar data original tetap aman
df_clean = df.copy()

# 1. Mengisi nilai kosong pada kolom numerik dengan median
numerik_cols = df_clean.select_dtypes(include=[np.number]).columns
for col in numerik_cols:
    if df_clean[col].isnull().any():
        if df_clean[col].isnull().all():
            df_clean[col] = df_clean[col].fillna(0)
        else:
            median_val = df_clean[col].median()
            df_clean[col] = df_clean[col].fillna(median_val)

# 2. Mengisi nilai kosong pada kolom objek/kategori dengan 'Unknown'
object_cols = df_clean.select_dtypes(include=['object']).columns
for col in object_cols:
    if df_clean[col].isnull().any():
        df_clean[col] = df_clean[col].fillna('Unknown')

# Memverifikasi hasil preprocessing
sisa_null = df_clean.isnull().sum().sum()
print(f"Jumlah kolom awal: {df.shape[1]} -> Jumlah kolom setelah diproses: {df_clean.shape[1]}")
print(f"Jumlah total nilai kosong (null) yang tersisa: {sisa_null}")
display(df_clean.head())
```

```python
# 1. Memeriksa sisa nilai kosong (null) di setiap kolom untuk memastikan tidak ada yang terlewat
missing_after = df_clean.isnull().sum()
missing_after_df = pd.DataFrame({"jumlah_null": missing_after, "persen_null": (missing_after / len(df_clean) * 100).round(2)})

print("--- Informasi Dataset Setelah Preprocessing ---")
print(f"Ukuran Baris & Kolom: {df_clean.shape}")
print(f"Jumlah Duplikat: {df_clean.duplicated().sum()}")
print("\n--- Kolom yang Sebelumnya Memiliki Null (Memastikan sudah bersih) ---")

# Menampilkan kolom-kolom yang sebelumnya memiliki null pada 'df' asli
kolom_null_sebelumnya = df.columns[df.isnull().sum() > 0]
display(missing_after_df.loc[kolom_null_sebelumnya].sort_values("jumlah_null", ascending=False).head(15))

print("\n--- Contoh Data Hasil Imputasi (5 baris acak) ---")
display(df_clean[kolom_null_sebelumnya].sample(5, random_state=42))
```
